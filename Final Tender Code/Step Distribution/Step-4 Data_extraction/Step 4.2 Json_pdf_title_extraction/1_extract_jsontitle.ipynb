```python
import json
import pandas as pd

# Function to load JSON data into a DataFrame
def load_json_to_dataframe(json_file):
    with open(json_file, 'r', encoding='utf-8') as f:
        data = json.load(f)
    
    # Create DataFrame from the JSON data
    df = pd.DataFrame(data)
    
    return df

# Path to your JSON file
json_file = 'extracted_pdf_data.json'

# Load the JSON data into a DataFrame
df = load_json_to_dataframe(json_file)

# Display the DataFrame
df
```

```python
import re
import pandas as pd

# Pre-compile start and end patterns to avoid recompiling during every function call
start_patterns = [
    r'INDEX\s*Name\s*of\s*Work:-',
    r'Tender\s*for\s*\SSupply\s*of',
    r'SUPPLY\s*AND\s*INSTALLATION\s*OF',
    r'supply\s*(,?\s*(and\s*)?)?\s*installation\s*(,?\s*(and\s*)?)?\s*(testing\s*(,?\s*(and\s*)?)?)?\s*commissioning\s*(of\s*)?',
    r'GLOBAL\s*TENDER\s*ENQUIRY\s*For',
    r'GLOBAL\s*TENDER\s*ENQUIRY\s*(?:\(\d\s*nd\s*Call\)|\(\d\s*st\s*Call\)|\(\d\s*rd\s*Call\))\s*For',
    r'Work\s*/Item\(s\)\s*Title',
    r'Rate\s*Contract\s*(document\s*)?for',
    r'Rate\s*Contract\s*of\s*Supply\s*and\s*Installation\s*of',
    r'Development\s*of',
    r'Tender\s*document\s*for\s*SITC\s*of',
    r'Rishikesh-249203',
    r'Brief\s*Description\s*of\s*Work\s*/\s*Services',
    r'Brief\s*Description\s*of\s*Goods\s*:',
    r'(E-?\s*)?TENDER\s*(DOCUMENT\s*)?FOR\s*SUPPLY\s*OF',
    r'E-\s*?TENDER\s*DOCUMENT',
    r'e\STender\s*For\s*SITC',
    r'Annexure-1\s*SPECIFICATIONS\s*OF',
    r'Tender\s*For\s*Setting',
    r'Tender\s*Notice\s*for\s*',
    r'NOTICE\s*INVITING\s*TENDER',
    r'Tender\s*Document\s*for\s*',
    r"Tender\s*For",
    r"Tender\s*Document",
    r'TENDER\s*ENQUIRY\s*For\s*',
    r'TENDER\s*ENQUIRY\s*DOCUMENT',
    r'TENDER\s*FOR',
    r'procurement\s*of',
    r'NOTICE\s*INVITING\s*e-TENDER',
    r'Notice\s*Inviting\s*Tender(\s*For\s*)?',
    r'TENDER\s*ENQUIRY\s*For(.*?)',
    r"E[-\s]*Tender\s*\(",
    r'INVITATION\s*FOR\s*QUOTATIONS',
    r'Supply\s*and\s*Installation\s*of',
    r'REQUEST\s*FOR\s*PROPOSAL',
    r'Technical\s*Specification\s*for',
    r'Limited\s*to\s*Registered',
    r'Auction\s*Brief\s*:'
]
end_patterns = [
    r'[Dd]epartment\s*[Oo]f\s*\w+',
    r"At\s*All\s*India\s*Institute\s*of\s*Medical\s*Sciences",
    r"All\s*India\s*Institute\s*of\s*Medical\s*Sciences\s*,?\s*[A-Z]*",
    r"AIIMS\s*,?\s*[A-Za-z]*",
    r"for the Department of\s*[A-Za-z\s]+",
    r'of\s*following\s*sizes',
    r'software',
    r'firm',
    r'cost',
    r'date',
    r'INDEX',
    r'page',
    r'\d\)',
    r'\.'
]

# Pre-compile end patterns once
end_pattern_combined = '|'.join(end_patterns)

# Function to extract tender titles
def extract_tender_titles(text, start_patterns, end_pattern_combined):
    # Limit the text to the first 550 words directly using slicing
    limited_text = ' '.join(text.split()[:550])
    
    for start_pattern in start_patterns:
        # Compile the regex once per start pattern
        pattern = re.compile(rf'(?i)({start_pattern}).*?({end_pattern_combined})', re.DOTALL)
        
        # Search for the pattern in the limited text
        match = pattern.search(limited_text)
        
        if match:
            # Clean and return the matched tender title
            return ' '.join(match.group(0).split())
    
    return ""

# Apply the function to the DataFrame
df['tender_title'] = df['content'].apply(lambda x: extract_tender_titles(x, start_patterns, end_pattern_combined))

# View the result
df[['file_name', 'tender_title']]
```

```python
# Save DataFrame to JSON file
json_file_path = 'tender_titles.json'  # Specify your file name
# Save DataFrame to JSON file
df.to_json(json_file_path, orient='records', lines=False, indent=4)

print(f"Data saved to {json_file_path}")
```
